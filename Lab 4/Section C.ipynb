Section C: Problem 2 – Count Pairs With a Given Sum

```python
import random
import time

def count_pairs_naive(nums, target):
    count = 0
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                count += 1
    return count


def count_pairs_optimized(nums, target):
    count = 0
    seen = {}
    for x in nums:
        complement = target - x
        if complement in seen:
            count += seen[complement]
        seen[x] = seen.get(x, 0) + 1
    return count


print("Naive version ")
for N in [5000, 50000]:
    random.seed(42)
    arr = [random.randint(1, 20000) for _ in range(N)]
    target = 20000

    start = time.time()
    result = count_pairs_naive(arr, target)
    end = time.time()
    print(f"N = {N}, pairs found = {result}, time = {end - start:.4f} s")

print()
print(" Optimimum version ")
random.seed(42)
N = 50000
arr = [random.randint(1, 20000) for _ in range(N)]
target = 20000

start = time.time()
result = count_pairs_optimized(arr, target)
end = time.time()
print(f"N = {N}, pairs found = {result}, time = {end - start:.4f} s")
```

When N increased from 5,000 to 50,000 (10 times), the execution time increased from 0.7824 seconds to 76.2584 seconds. This is approximately 97 times slower. ## Q9. What category of change made this faster? A smarter algorithm, a smarter data structure, or both? Explain.

Both a smarter algorithm and a smarter data structure made this approach faster. The optimized method uses a dictionary to store previously visited numbers and their frequencies. For every number, it calculates the required complement (target - x) and checks whether it already exists in the dictionary. Dictionary lookup takes approximately O(1) time on average, avoiding the need to check every pair using nested loops. This reduces the overall time complexity from O(N²) to O(N), making the optimized solution significantly faster.
