# Section D: Problem 3 – K-th Smallest Pairwise Sum

```python
import random
import time

def kth_smallest_sum_naive(arr, k):
    sums = []
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            sums.append(arr[i] + arr[j])
    sums.sort()
    return sums[k - 1]


def count_pairs_leq(arr, v):
    n = len(arr)
    count = 0
    left, right = 0, n - 1
    while left < right:
        if arr[left] + arr[right] <= v:
            count += right - left
            left += 1
        else:
            right -= 1
    return count


def kth_smallest_sum_optimized(arr, k):
    arr = sorted(arr)
    n = len(arr)
    lo = arr[0] + arr[1]
    hi = arr[n - 2] + arr[n - 1]
    while lo < hi:
        mid = (lo + hi) // 2
        if count_pairs_leq(arr, mid) >= k:
            hi = mid
        else:
            lo = mid + 1
    return lo



random.seed(42)
N = 5000
arr = [random.randint(1, 100000) for _ in range(N)]
K = 1000000

print(" Naive version ")
start = time.time()
result_naive = kth_smallest_sum_naive(arr, K)
end = time.time()
print(f"N = {N}, K = {K}, result = {result_naive}, time = {end - start:.4f} s")

print()
print(" Optimimum version")
start = time.time()
result_opt = kth_smallest_sum_optimized(arr, K)
end = time.time()
print(f"N = {N}, K = {K}, result = {result_opt}, time = {end - start:.4f} s")

```

# Section D: Problem 3 – K-th Smallest Pairwise Sum

## Step 1: Naive Version – Observation

The naive approach generates all possible pairwise sums, stores them in a list, sorts the entire list, and then selects the K-th smallest element.

For N = 5,000, the total number of possible pairs is:

N × (N - 1) / 2 = 5000 × 4999 / 2 = 12,497,500 pairs.

This requires a large amount of memory to store all the sums. Sorting such a large list also takes considerable execution time, making this approach inefficient for large inputs.

## Step 2: Hint 1 – Binary Search on Answer

Instead of generating and sorting all pairwise sums, we can calculate how many pairwise sums are less than or equal to a given value V. Using this count, we can determine whether the K-th smallest sum is greater than or equal to V or lies below it. This helps us narrow down the answer efficiently using binary search.

## Step 3: Hint 2 – Two Pointer Technique

After sorting the array, we use two pointers, one at the beginning and another at the end. If the sum of the elements at these positions is less than or equal to V, then all pairs between the left pointer and the right pointer also satisfy the condition. Therefore, we can count them together without checking each pair individually. Otherwise, we decrease the right pointer.

This reduces the counting operation to O(N) time.

## Step 4: Hint 3 – Binary Search Over Possible Sums

We first sort the array and determine the minimum and maximum possible pairwise sums. Then, we apply binary search over this range. For every middle value, we count how many pairwise sums are less than or equal to it using the two-pointer technique. If the count is at least K, we search in the smaller half; otherwise, we search in the larger half. We continue until we find the K-th smallest pairwise sum.

## Q10. Compare your naive and optimized times. What made the biggest difference: avoiding storing every sum, or searching over answers instead of generating them directly?

Both avoiding the storage of every pairwise sum and searching over possible answers contributed to the improvement.
